# APC closed-loop behavior verification

One contracted Paper 1 verification at Δh = 90 m and seed 42. Scientific parameter drift is rejected. A successful run writes `verification_report.json` below `artifacts/experiments/apc_closed_loop_verification/<run_id>/`.


```python
from pathlib import Path
import json
import pandas as pd
from IPython.display import Image, display

from loraphymac.experiments.apc_closed_loop_verification.workflow import (
    default_apc_closed_loop_config,
    run_apc_closed_loop_verification,
    verification_report_path,
)
from loraphymac.experiments.common.single_run_diagnostics import (
    generate_single_run_figures,
    generate_single_run_tables,
)
```

```python
config = default_apc_closed_loop_config()
# Only operational fields such as run_id/output_dir/print flags should be changed.
display(config)
```

```python
result = run_apc_closed_loop_verification(config)
figure_paths = generate_single_run_figures(result)
table_paths = generate_single_run_tables(result)

for figure_name, paths in figure_paths.items():
    preview = next((path for path in paths if path.suffix.lower() in {".png", ".jpg"}), None)
    print(f"{figure_name}: {', '.join(str(path.resolve()) for path in paths)}")
    if preview is not None:
        display(Image(filename=str(preview)))

for table_name, path in table_paths.items():
    print(f"{table_name}: {path.resolve()}")
    display(pd.read_csv(path))

report = verification_report_path(result)
display(json.loads(report.read_text(encoding="utf-8")))
print(f"Verification output: {Path(result.resolved_config['output_dir']).resolve()}")
```
