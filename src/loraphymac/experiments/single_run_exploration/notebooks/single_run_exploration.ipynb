# LoRaPhyMAC single-run exploration

Flexible one-run sandbox with central configuration validation but no publication contract. Edit the override block, then run all cells. Outputs go to `artifacts/experiments/single_run_exploration/<run_id>/`.


```python
from pathlib import Path
import pandas as pd
from IPython.display import Image, display

from loraphymac.experiments.single_run_exploration import (
    default_exploration_config,
    generate_single_run_figures,
    generate_single_run_tables,
    run_single_exploration,
)
```

```python
config = default_exploration_config()
config.update({
    # Freely change exploratory parameters here.
    # "run_id": "exploration_multichannel_001",
    # "relay_tx_ch_ids": [1, 2, 3, 4],
})
display(config)
```

```python
result = run_single_exploration(config)
figure_paths = generate_single_run_figures(result)
table_paths = generate_single_run_tables(result)

for figure_name, paths in figure_paths.items():
    preview = next((path for path in paths if path.suffix.lower() in {".png", ".jpg"}), None)
    print(f"{figure_name}: {', '.join(str(path.resolve()) for path in paths)}")
    if preview is not None:
        display(Image(filename=str(preview)))

for table_name, path in table_paths.items():
    print(f"{table_name}: {path.resolve()}")
    display(pd.read_csv(path))

print(f"Output directory: {Path(result.resolved_config['output_dir']).resolve()}")
```
