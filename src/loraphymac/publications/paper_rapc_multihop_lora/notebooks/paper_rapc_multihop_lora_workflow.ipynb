# R-APC Multi-Hop LoRa multi-experiment publication workflow

Notebook ini membaca artefak tervalidasi dari terrain sweep dan closed-loop verification, kemudian menyusun tabel serta grafik publikasi. Notebook tidak menjalankan simulasi.


```python
from IPython.display import Image, Markdown, display
from loraphymac.publications.paper_rapc_multihop_lora.analysis import publication_paths
from loraphymac.publications.paper_rapc_multihop_lora.workflow import (
    run_publication, validate_publication_source
)
```

```python
spec, destination, evidence = validate_publication_source()
display({
    "status": "READY",
    "terrain_sweep": evidence.terrain_sweep.paths.experiment_dir,
    "closed_loop_verification": (
        evidence.closed_loop_verification.experiment_dir
        if evidence.closed_loop_verification is not None else None
    ),
    "publication_output": destination.publication_dir,
    "terrain_run_count": evidence.terrain_sweep.status["passed_run_count_on_disk"],
})
```

```python
publication_summary = run_publication()
display(publication_summary)
figure_dir = publication_paths().figures

display(Markdown("## Terrain-sweep results (300 runs)"))
for figure in sorted(figure_dir.glob("figure_0[1-6]_*.png")):
    display(Image(filename=str(figure)))

display(Markdown("## Closed-loop APC verification"))
closed_loop_figures = [
    *sorted(figure_dir.glob("figure_0[7-9]_*.png")),
    *sorted(figure_dir.glob("figure_10_*.png")),
]
for figure in closed_loop_figures:
    display(Image(filename=str(figure)))
```
